```python
def evenodd(x: int)->str:
    if(x%2==0):
        return "Even"
    else:
        return "Odd"
x=int(input("Enter a number"))
print(evenodd(x))
```

```python
x=int(input("enter x :"))
y=int(input("enter y :"))
z=int(input("enter z :"))

def cal_sum_product(x,y,z):
    ss=x+y+z
    pp=x*y*z
    return ss,pp

s,p=cal_sum_product(x,y,z)
print(s,p)
```
